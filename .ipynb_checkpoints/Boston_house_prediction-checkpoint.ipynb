### 6.Outlier Detection

```python
plt.figure(figsize=(20,12))
for i,col in enumerate(df):
    plt.subplot(4,4,i+1)
    sns.boxplot(x=df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
# calculate outlier counts
Q1= df.quantile(0.25)
Q3= df.quantile(0.75)

IQR= Q3 - Q1

lower_bound= Q1-1.5*IQR
upper_bound= Q3+1.5*IQR

outlier_count=((df<lower_bound) | (df>upper_bound)).sum()
print(outlier_count)
```

```python
outlier_percentage = (outlier_count / len(df)) * 100

print(outlier_percentage)
```

```python
outlier_cols = ['CRIM', 'ZN', 'RM', 'DIS', 'PTRATIO', 'B', 'LSTAT']

for col in outlier_cols:
    
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    
    IQR = q3 - q1
    
    lb = q1 - 1.5 * IQR
    ub = q3 + 1.5 * IQR
    
    outliers = ((df[col] < lb) | (df[col] > ub)).sum()
    
    print(f"\n{col}")
    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", IQR)
    print("Lower Bound:", lb)
    print("Upper Bound:", ub)
    print("Outliers:", outliers)
```

**observations:**
- IQR method identified potential outliers in CRIM, ZN, RM, DIS, PTRATIO, B, and LSTAT.
- CRIM, ZN, and B have a relatively high number of potential outliers.
- DIS and LSTAT have comparatively fewer potential outliers.
- CHAS is a binary variable, so its flagged values are not treated as conventional outliers.
- MEDV is the target variable, so we will not modify its outliers.
- We will use IQR capping for the selected feature columns instead of deleting rows.

```python
# IQR capping method
for col in outlier_cols:
    
    q1 = np.percentile(df[col], 25)
    q3 = np.percentile(df[col], 75)

    IQR = q3 - q1

    lb = q1 - 1.5 * IQR
    ub = q3 + 1.5 * IQR

    df[col] = df[col].clip(lb, ub)
```

```python
plt.figure(figsize=(20,12))
for i,col in enumerate(df):
    plt.subplot(4,4,i+1)
    sns.boxplot(x=df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
### 7.Train-Test-Split
x=df.drop('MEDV',axis=1)
y=df['MEDV']
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,
                                               test_size=0.2,
                                               random_state=42)
print("x_train:", x_train.shape)
print("x_test:", x_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

**observation:**
- Dataset was split into 80% training and 20% testing data.
- Training data contains 404 observations, while testing data contains 102 observations.
- There are 13 independent features and 1 dependent variable (MEDV).
- The test data will be used later to evaluate the model's performance on unseen data.

### 8.Feature Scalilng

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler
```

```python
X_train_scaled = scaler.fit_transform(x_train)
X_test_scaled = scaler.transform(x_test)
print(X_train_scaled.shape)
print(X_test_scaled.shape)
```

### 9.Feature Selection

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
rfe = RFE(
    estimator=model,
    n_features_to_select=8
)

rfe.fit(X_train_scaled, y_train)
```

```python
selected_features = x_train.columns[rfe.support_]

print("Selected Features:")
print(selected_features)
```

```python
X_train_selected = X_train_scaled[:, rfe.support_]
X_test_selected = X_test_scaled[:, rfe.support_]
print(X_train_selected.shape)
print(X_test_selected.shape)
```

### 10.Model Building

#### 1. Linear Regression Model

```python
from sklearn.linear_model import LinearRegression
lr_model=LinearRegression()
lr_model
```

```python
lr_model.fit(X_train_selected, y_train)
```

**# model prediction**

```python
y_pred= lr_model.predict(X_test_selected)
y_pred
```

**# model evaluation**

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("MAE :", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R²  :", r2)
```

**observation:**
- The Linear Regression model achieved an MAE of 3.12, RMSE of 4.77, and R² of 0.69. The model explains approximately 69% of the variation in house prices, with an average absolute prediction error of around $3,120.
